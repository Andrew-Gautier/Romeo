```python
import torch
from torchviz import make_dot
from LSTMClassifer import LSTMClassifier, VOCAB_SIZE, EMBEDDING_SIZE, LSTM_NODES, device


# Create a sample input to trace through the model
batch_size = 2  # Small batch for visualization
sample_input = torch.randint(0, VOCAB_SIZE, (batch_size, 98)).to(device)  # Shape: [batch_size, sentence_length]

# Initialize the model (you'll need to provide the pretrained_weights)
# For visualization purposes, we can create dummy weights
dummy_weights = torch.randn(VOCAB_SIZE, EMBEDDING_SIZE).to(device)

model = LSTMClassifier(
    vocab_size=VOCAB_SIZE,
    embedding_dim=EMBEDDING_SIZE,  
    hidden_dim=LSTM_NODES,
    output_dim=50,  
    n_layers=2,
    batch_first=True,
    bidirectional=True,
    dropout=0.5,
    pretrained_weights=dummy_weights
).to(device)

# Set model to evaluation mode
model.eval()

# Forward pass to capture the computation graph
with torch.no_grad():
    output = model(sample_input)

# Create the visualization
dot = make_dot(output, params=dict(model.named_parameters()))

# Save the visualization
dot.render('lstm_selfattention_architecture', format='png', cleanup=True)
print("Model architecture visualization saved as 'lstm_selfattention_architecture.png'")

# Optional: Also create a text summary
print("\nModel Summary:")
print(model)
print(f"\nInput shape: {sample_input.shape}")
print(f"Output shape: {output.shape}")
```
